```python
import numpy as np

def is_on_curve(x, y, N):
    tol = 1 / N
    return abs((x**2 + y**2)**3 - 4 * x**2 * y**2) <= tol

def crosses_curve(x, y, angle, L, N):
    # Endpoints of a needle with a given center
    dx = (L/2) * np.cos(angle)
    dy = (L/2) * np.sin(angle)
    x1, y1 = x + dx, y + dy
    x2, y2 = x - dx, y - dy

    # Sample points along the needle
    ts = np.linspace(0, 1, N)
    xs = x1 * ts + x2 * (1 - ts)
    ys = y1 * ts + y2 * (1 - ts)

    # Check if any point is on the curve
    on_curve = is_on_curve(xs, ys, N)
    return np.any(on_curve)
```

```python
def estimate_probability(L, N):
    count = 0
    for _ in range(N):
        x, y = np.random.uniform(-1, 1, 2)
        alpha = np.random.uniform(0, 2 * np.pi)
        if crosses_curve(x, y, alpha, L, N):
            count += 1
    return count / N
```

```python
lengths = [1/10, 1/5, 1/4, 1/3, 1/2, 1]
print(f"{'Needle Length':>15} | {'Probability of Crossing':>25}")
print("-"*45)
for L in lengths:
    prob = estimate_probability(L, 10000)
    print(f"{L:15.4f} | {prob:25.4f}")
```
