```python

import numpy as np

# Given function
def f(x):
    return np.exp(x**5)

# Given true value
true_value = 2.0949681713212

# Integration limits
a, b = -1, 1
```

## a) Midpoint

```python

def midpoint_rule(f, a, b, N):
    dx = (b - a) / N
    x = np.array([a + dx / 2 + i * dx for i in range(N)])
    return dx * sum(f(x))

result = midpoint_rule(f, a, b, 100)
error = abs(result - true_value)
print(f"Midpoint\nResult: {result:.12f}\nError:  {error:.12f}")
```

## b) Simpson 1/3

```python
def simpson_one_third(f, a, b, N):
    dx = (b - a) / (N - 1)
    x = np.array([a + i * dx for i in range(N)])
    weights = np.ones(N)
    weights[1::2] = 4  # All odd indices get weight 4
    weights[2::2] = 2  # All even indices (except first/last) get weight 2
    
    return (dx / 3) * np.sum(weights * f(x))

result = simpson_one_third(f, a, b, 101)
error = abs(result - true_value)
print(f"Simpson 1/3\nResult: {result:.12f}\nError:  {error:.12f}")
```

## c) Simpson 3/8

```python
def simpson_three_eighths(f, a, b, N):
    dx = (b - a) / N
    weights = np.ones(N)
    weights[1:-1] = 3  # Second to second-last terms
    
    terms = 0
    for i in range(N):
        terms += weights[i] * f(a + i * dx)
    
    return (3 * dx / 8) * terms

result = simpson_three_eighths(f, a, b, 101)
error = abs(result - true_value)
print(f"Simpson 3/8\nResult: {result:.12f}\nError:  {error:.12f}")
```

## d) Gaussian Quadrature

```python
def gaussian_quadrature(f, a = -1, b = 1, N = 5):
    # Pre-computed Gauss-Legendre points and weights for N=5
    x = np.array([
        -0.90618,
        -0.538469,
        0,
        0.538469,
        0.90618
    ])
    
    w = np.array([
        0.236927,
        0.478629,
        0.568889,
        0.478629,
        0.236927
    ])
    
    # No need to transform since a,b is already [-1,1]
    return np.sum(w * f(x))

result = gaussian_quadrature(f, a, b)
error = abs(result - true_value)
print(f"Gaussian Quadrature\nResult: {result:.12f}\nError:  {error:.12f}")
```

## e) Monte Carlo

```python
def monte_carlo(f, a, b, N):
    x = np.random.uniform(a, b, N)
    return (b - a) * np.mean(f(x))

result = monte_carlo(f, a, b, 1000)
error = abs(result - true_value)
print(f"\nMonte Carlo\nResult: {result:.12f}\nError:  {error:.12f}")
```
